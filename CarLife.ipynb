# A Day in the Life of a Car 🚗⛽

This notebook tells the story of a car's daily life through data.
We will follow a car's mileage over a week and see how numbers
can turn into insights.

> *"Numbers don’t lie, but they tell a story."*


## Why Car Data?

Cars are part of daily life—they travel, consume fuel, and need maintenance.  
In this notebook, we'll use **fictional car mileage data** to:
- Record daily travel distance
- Calculate fuel efficiency
- Visualize the data
- Interpret insights


```python
# Mileage (km) for a week (fictional data)
mileage = [50, 75, 30, 100, 60, 0, 80]

mileage
```

## Meet the Data

Each number represents **the distance driven by the car in a day (km)**.  
Numbers alone don't tell the full story. Let's ask:  
*"How much did the car travel this week?"*


```python
# Total and average mileage
total_mileage = sum(mileage)
average_mileage = total_mileage / len(mileage)

total_mileage, average_mileage
```

## What Do the Numbers Say?

- **Total mileage**: How far the car traveled this week  
- **Average mileage**: Daily driving consistency  

Graphs can make this even clearer.


```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

plt.plot(days, mileage, marker='o', color='blue')
plt.title("Car Mileage Over a Week")
plt.xlabel("Day")
plt.ylabel("Distance (km)")
plt.grid(True)
plt.show()
```

## Real-World Applications 🚘
